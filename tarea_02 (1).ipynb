```python
# Tarea 02 - Narración Gráfica

# Problema 1: Clasificador de temperatura
# Problema 1: Clasificador de temperatura

# Recorremos los números desde -30 hasta 50
for temperatura in range(-30, 51):

    # Si la temperatura es menor que 0
    if temperatura < 0:
        print(temperatura, "Muy frío")

    # Si la temperatura está entre 0 y 24
    elif temperatura <= 24:
        print(temperatura, "Templado")

    # Si la temperatura es mayor o igual a 25
    else:
        print(temperatura, "Caluroso")
```

```python
# Problema 2: Números especiales del robot
# Problema 2: Números especiales del robot

# Recorremos los números del 1 al 100
for numero in range(1, 101):

    # Primero verificamos si el número es divisible por 5
    if numero % 5 == 0:

        # Si además es divisible por 10, mostramos "Bip Bip"
        if numero % 10 == 0:
            print(numero, "Bip Bip")

        # Si no es divisible por 10, mostramos "Bip"
        else:
            print(numero, "Bip")
```

```python
# Problema 3: Detector de puntaje gamer
# Problema 3: Detector de puntaje gamer

# Pedimos al usuario los dos puntajes
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Sumamos los dos puntajes
suma = puntaje1 + puntaje2

# Evaluamos el nivel según el resultado
if suma < 50:
    print("Resultado: Nivel principiante")

elif suma < 80:
    print("Resultado: Nivel intermedio")

else:
    print("Resultado: Nivel pro")
```

```python
# Problema 4: Calculador de perfil musical

def calcular_perfil(edad, le_gusta_musica, le_gusta_bailar):

    # Verificamos si cumple todas las condiciones para ser una estrella de la fiesta
    if edad >= 18 and le_gusta_musica and le_gusta_bailar:
        return "Eres una estrella de la fiesta"

    # Verificamos si le gusta la música, pero no bailar
    elif le_gusta_musica and not le_gusta_bailar:
        return "Te gusta la música, pero no tanto bailar"

    # Para cualquier otro caso
    else:
        return "Tienes tu propio estilo"


# Pedimos los datos al usuario
edad = int(input("Ingresa tu edad: "))
le_gusta_musica = input("¿Te gusta la música? (si/no): ").lower() == "si"
le_gusta_bailar = input("¿Te gusta bailar? (si/no): ").lower() == "si"

# Mostramos el resultado
print(calcular_perfil(edad, le_gusta_musica, le_gusta_bailar))
```
